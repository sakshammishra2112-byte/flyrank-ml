## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
## Answer

My project is a ranking and scoring machine learning task.

The goal is to assign every content page a refresh opportunity score and rank the pages from highest to lowest priority. Pages with higher scores should be reviewed first because they are more likely to benefit from updates.

A scoring approach is more suitable than simple classification because every page can have a different level of opportunity instead of only belonging to categories like "refresh" or "don't refresh".
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
The target is the refresh opportunity score for each content page.

Since there is no direct label showing whether a page should be refreshed, the project uses observable search and content signals as a proxy. Signals such as traffic trend, impressions, clicks, CTR and content freshness help estimate which pages deserve review.

The model supports human decisions rather than making automatic publishing decisions.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
A fixed rule cannot capture the interaction between multiple signals such as traffic trend, CTR, impressions, clicks and content freshness.

Some pages may have decreasing traffic but increasing CTR, while others may have high impressions but low engagement. Machine learning can learn these complex relationships and produce better rankings than a single manually written rule.

The model provides decision support and should always be reviewed by humans before taking action.
```
